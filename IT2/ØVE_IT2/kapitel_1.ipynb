```python
# Oppgave 3

maaneder = "JanFebMarAprMaiJunJulAugSepOktNovDes"

def hent_maandernummer(maanedsnummer):
    if maanedsnummer < 1 or maanedsnummer > 12:
        return "Ugyldig data, prøv igjen."
    else:
        startindeks = (maanedsnummer - 1) * 3
        forkortelse = maaneder[startindeks:startindeks + 3]
        return forkortelse

def main():
    maanedsnummer = int(input("Skriv inn et tall fra 1 til 12: "))
    maanedsforkortelse = hent_maandernummer(maanedsnummer)
    print(f" Forkortelsen til månednummer er {maanedsforkortelse}")

main()
   
```
