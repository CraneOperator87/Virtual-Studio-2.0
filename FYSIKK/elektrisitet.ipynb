```python
# Parallel kobling

def main():
    R1 = 20 #ohm
    R2 = 50 #ohm
    Vei1 = R1 + R2
    print(f"Total motstand: {Vei1} ohm")
    R3 = 10 #ohm
    R4 = 30 #ohm
    R5 = 30 #ohm
    parallel = 1 / (1/R4 + 1/R5)
    print(f"motstand parallel med R3 og R4: {parallel} ohm")
    Vei2 = R3 + parallel
    R6 = 50 #ohm
    R7 = 50 #ohm
    print(f"Total motstand: {1/((1/Vei1)+(1/Vei2)+(1/R7)) + 40} ohm")

main()
```
