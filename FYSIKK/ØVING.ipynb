```python
# Konstanter
B = 2.18 # i attojoule
h = 6.63e-34
c = 3e8
 
# 1) Lag en oversikt over energitilstander (L)
E = []
for i in range(1,11):
    E.append(-B/i**2)
 
# 2) Lag en tabell for alle mulige overganger (M/V)
Etabell = {}
for i in range(0,10):
    for j in range(i+1, 10):
        Etabell[(i+1, j+1)] = E[j] - E[i]
 
# 3) Hent data fra brukeren (L)
valg = input("Frekvens (f) eller bølgelengde (l)?")
verdi = float(input("Oppgi verdien til størrelsen i SI-enheter"))
if valg=="f":
    Eo = h*verdi/1e-18
elif valg=="l":
    Eo = h*c/verdi/1e-18
 
# 4) Slå opp i tabellen (V)
beste_baner = None
min_avvik = float('inf')
for overgang in Etabell:
    avvik = abs(Eo - Etabell[overgang])
    if avvik < min_avvik:
        min_avvik = avvik
        beste_baner = overgang
 
# 5) Vis resultatene på en oversiktlig måte (L)
print("Fotonet svarer til", round(Eo, 3), "aJ")
print("Den nærmeste overgangen i hydrogen er", beste_baner)
print("Denne overgangen har energien lik", round(Etabell[beste_baner], 3), "aJ")
 
 
 
 
 
 
 

```

```python
def main():
    # Konstanter
    B = 2.18E-18
    H = 6.63E-34
    c = 3E8
    
    # Energinivåer
    E1 = -0.8234E-18
    E2 = -0.4861E-18

    # Bohrs formel
    # E = - B * ((-1/n1**2)-(-1(n2**2))
    deltaE = E2 - E1
    print(f"Delta E: {deltaE:.2E} J")

    # lambda = c/f
    lambda1 = (c * H) / deltaE
    print(f"Lambda: {lambda1:.2E} m")
main()
```

```python
# Fisjon
def main():
    # 2H + 2H -> 3He + 1n
    u = 1.66e-27
    c = 3e8
    m_2H = 2.014102
    m_3He = 3.016029 
    m_n = 1.008665 

    # Frigjort energi: E_0 = (m_før - m_etter)* u * c**2
    E = (2 * m_2H - m_3He - m_n) * u * c**2
    print(f"Frigjort energi: {E:.2E} J")
main()
```

```python
# Fusjon
def main():
    # Konstanter 
    c = 3e8
    u = 1.66e-27 
    # Masser
    m_n = 1.008665 # Nøytron
    m_p = 1.007276 # Proton
    m_e = 0.000549 # Elektron
    # Formel
    # E_0 = (m_før - m_etter) * u * c**2
    E = (m_n - (m_p + m_e)) * u * c**2
    print(f"Frigjort energi: {E:.2E} J")

main()
```
