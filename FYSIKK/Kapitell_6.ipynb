```python
# Oppgave 6.9

def main():
    # Kosntanter : 
    a = 2.90e-3

    # Verdier
    Rigel = 12000
    Betelgeuse = 3000

    # Bølgetopp * T = a
    # -> a/T = bølgetopp

    # Rigler
    bølgetopp_Rigel = a /Rigel
    bølgetopp_Betelgeuse = a /Betelgeuse
    print(f" Bølgetoppene til Rigel: {bølgetopp_Rigel:.2e} m, og Betelgeuse: {bølgetopp_Betelgeuse:.2e} m")

main()
```
