```python
def main():
    Pa = 1.740E17
    Pb = Pa*8
    taTotal = 10E10
    maTotal = 1.980E30
    mbTotal  = maTotal*20
    tbTotal = (Pa/Pb)*(maTotal/mbTotal)*(taTotal) 

    print(f"tbTotal = {tbTotal:.2e}")
main()
```

```python
def main():
    mSun = 1.980E30
    mStar = mSun * 20
    LifetimeStar = (mSun/mStar) ** 3

    print(f"LifetimeStar = {LifetimeStar:.2e}")

    LifetimeSun = 10e9
    LifeStar = LifetimeSun * LifetimeStar

    print(f"LifeStar = {LifeStar:.2e}")

main()
```
