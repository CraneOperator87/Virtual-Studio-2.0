```python
# 8.16
def main():
    U238 = 238.050788            # Uran 238
    Th234 = 234.043601           # Thorium 234
    He4 = 4.002603               # Helium 4
    energi_før = U238                      # Vi finner energien før fisjonen
    energi_etter = Th234 + He4             # Vi finner energien etter fisjonen
    differanse = energi_før - energi_etter # Vi finner differansen
    c = 3e8                                # Lysets hastighet
    deltaE = differanse * 1.66e-27 * c**2  # delta E = differansen i massen ganget med c^2 gir oss energien i Joule som ble frigjort
    print(f"Delta E = {deltaE:.2e} J ")

main()
```

```python
# 8.42
def main():
    c = 3e8
    ø = 1.66e-27
    U235 = 235.043928
    Sr88 = 87.905612
    Xe136 = 135.907219
    n = 1.008665
    energi_før = U235 + n
    energi_etter = Sr88 + Xe136 + (12*n)
    differanse = energi_før - energi_etter
    deltaE = differanse * ø * c**2
    print(f"Delta E = {deltaE:.2e} J ")

main()
```
