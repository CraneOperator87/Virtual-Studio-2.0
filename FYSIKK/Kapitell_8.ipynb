```python
# 8.57b
# 14N + energi -> 13C + 1p

def main():
    # Konstanter:
    c = 3e8
    u = 1.66e-27 
    # Masser: 
    m_14N = 14.003074
    m_13C = 13.003355 
    m_1p = 1.007825
    # Ligning:
    # 14 N + energi -> 13C + 1p
    # m_14N*c**2 + energi = m_13C*c**2 + m_p*c**2
    # energi = ((m_13C*c**2 + m_p*c**2) - m_14N*c**2)
    # energi = (etter - før) * u * c ** 2
    E = ((m_13C + m_1p) - m_14N) * u * c ** 2
    print(f"Energi frigjort i reaksjonen: {E:.2e} J.")
main()
```
